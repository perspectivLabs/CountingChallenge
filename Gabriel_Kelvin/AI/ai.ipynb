```python
# Imports
import os
import cv2
import numpy as np
import matplotlib.pyplot as plt
from ultralytics import YOLO

# Load pre-trained YOLOv8 model (no training needed)
def load_model():
    # Using YOLOv8n for faster processing
    model = YOLO('yolov8n.pt')
    return model

# Enhanced counting function using YOLOv8's instance segmentation
def count_objects(image_path, model, conf_threshold=0.25):
    # Read image
    img = cv2.imread(image_path)
    original = img.copy()
    
    # Run inference with YOLO
    results = model(img, conf=conf_threshold)
    
    # Get detected objects from specific classes
    # YOLOv8 classes from COCO dataset - 76: scissors, 77: teddy bear (shape similar to screws)
    # 47: cup, 74: mouse, 75: remote (can detect small electronic parts)
    # Add 39: bottle, 41: cup, 42: fork, 43: knife, 44: spoon
    target_classes = [39, 41, 42, 43, 44, 47, 74, 75, 76, 77]
    
    # Filter results for our target classes
    detected_items = []
    overlay = original.copy()
    
    # Process all detected objects
    for r in results:
        boxes = r.boxes
        for box in boxes:
            # Check confidence
            conf = box.conf.cpu().numpy()[0]
            if conf < conf_threshold:
                continue
                
            # Get class
            cls = int(box.cls.cpu().numpy()[0])
            
            # Check if class is of interest or use all small objects
            if cls in target_classes or box.xywh.cpu().numpy()[0][2] * box.xywh.cpu().numpy()[0][3] < 10000:
                # Get box coordinates
                x1, y1, x2, y2 = box.xyxy.cpu().numpy()[0]
                x1, y1, x2, y2 = int(x1), int(y1), int(x2), int(y2)
                
                # Calculate object area
                object_area = (x2 - x1) * (y2 - y1)
                
                # Filter by size (to focus on smaller objects like screws)
                if object_area < 15000:  # Adjust this threshold based on your images
                    detected_items.append((x1, y1, x2, y2))
                    # Draw rectangle
                    cv2.rectangle(overlay, (x1, y1), (x2, y2), (0, 255, 0), -1)
    
    # If nothing detected with class filtering, use edge detection as fallback
    if len(detected_items) == 0:
        gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
        blurred = cv2.GaussianBlur(gray, (5, 5), 0)
        edges = cv2.Canny(blurred, 50, 150)
        
        # Find contours
        contours, _ = cv2.findContours(edges, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
        
        for contour in contours:
            area = cv2.contourArea(contour)
            if 100 < area < 5000:  # Filter by area to exclude noise and large objects
                x, y, w, h = cv2.boundingRect(contour)
                detected_items.append((x, y, x+w, y+h))
                cv2.rectangle(overlay, (x, y), (x+w, y+h), (0, 255, 0), -1)
    
    # Apply DBSCAN clustering to group nearby detections
    if len(detected_items) > 0:
        # Convert boxes to centroids
        centroids = np.array([[int((x1+x2)/2), int((y1+y2)/2)] for x1, y1, x2, y2 in detected_items])
        
        # Normalize coordinates for DBSCAN
        from sklearn.cluster import DBSCAN
        from sklearn.preprocessing import StandardScaler
        X = StandardScaler().fit_transform(centroids)
        
        # Apply DBSCAN
        db = DBSCAN(eps=0.3, min_samples=1).fit(X)
        labels = db.labels_
        
        # Count unique clusters
        n_clusters = len(set(labels)) - (1 if -1 in labels else 0)
        count = n_clusters
    else:
        count = 0
    
    # Blend original and overlay
    final_img = cv2.addWeighted(original, 0.7, overlay, 0.3, 0)
    
    return final_img, count

# Helper function to show image
def show_img(img, title='', size=(12, 12)):
    plt.figure(figsize=size)
    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    plt.imshow(img_rgb)
    plt.title(title)
    plt.axis('off')
    plt.show()

# Main function
def main():
    # Paths
    data_path = '../Data/'  # Update path if needed
    folders = ['Screws', 'ScrewAndBolt']
    
    # Load model once
    model = load_model()
    
    # Process images
    for folder in folders:
        folder_path = os.path.join(data_path, folder)
        if not os.path.exists(folder_path):
            print(f"Folder not found: {folder_path}")
            continue
            
        print(f"\n🔵 Processing Folder: {folder}")
        for filename in os.listdir(folder_path):
            if filename.lower().endswith(('.jpg', '.png')):
                image_path = os.path.join(folder_path, filename)
                result_img, count = count_objects(image_path, model)
                print(f"🖼️ Image: {filename} | Detected Items: {count}")
                show_img(result_img, title=f"{filename} - Items: {count}")

if __name__ == "__main__":
    main()
```
