# Non-AI Object Counting — OpenCV

This notebook runs a classical computer-vision pipeline using thresholding, morphology, connected components and watershed. No neural network is used.

## 1. Install dependencies

```python
from pathlib import Path
import os, subprocess, sys
PROJECT_ROOT = Path.cwd()
if not (PROJECT_ROOT / "evaluate_counts.py").exists():
    PROJECT_ROOT = PROJECT_ROOT.parent
NON_AI_DIR = PROJECT_ROOT / "Non_AI"
subprocess.run([sys.executable, "-m", "pip", "install", "-r", str(NON_AI_DIR / "requirements.txt")], check=True)
```

## 2. Set paths

```python
DATASET = Path(os.environ.get("COUNTING_DATASET", str(PROJECT_ROOT.parent / "data"))).resolve()
OUTPUT = NON_AI_DIR / "final_output"
GROUND_TRUTH = PROJECT_ROOT / "ground_truth.csv"
```

## 3. Run the pipeline

```python
subprocess.run([sys.executable, str(NON_AI_DIR / "non_ai_counting.py"), "--input", str(DATASET), "--output", str(OUTPUT)], cwd=PROJECT_ROOT, check=True)
subprocess.run([sys.executable, str(PROJECT_ROOT / "evaluate_counts.py"), "--predictions", str(OUTPUT / "predictions.csv"), "--ground-truth", str(GROUND_TRUTH), "--output", str(OUTPUT / "evaluation.csv")], cwd=PROJECT_ROOT, check=True)
```

## 4. Evaluation

Evaluation uses only `ground_truth.csv`; provisional counts in `ground_truth_plan.csv` are excluded.
